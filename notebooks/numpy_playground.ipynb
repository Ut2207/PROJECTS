```python
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 500)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 500)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 10, 500)
y = np.sin(5 * x)

plt.plot(x, y)
plt.title("Higher Frequency")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 10, 500)

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")

plt.legend()
plt.grid(True)
plt.show()
```

```python
t = np.linspace(0, 20 * np.pi, 5000)

x = np.sin(3 * t)
y = np.sin(4 * t)

plt.figure(figsize=(6, 6))
plt.plot(x, y)
plt.axis("equal")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("dark_background")

t = np.linspace(0, 40*np.pi, 8000)

x = np.sin(3*t)
y = np.sin(4*t + np.pi/2)

plt.figure(figsize=(6,6))
plt.plot(x, y, color="#c4a7e7", linewidth=1)

plt.axis("equal")
plt.axis("off")

plt.show()
```
